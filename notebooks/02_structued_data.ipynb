```python
def get_attendance(
    employee_id,
    start_date,
    end_date
):
    """
    Return attendance records for an employee
    within a specified date range.
    """

    start_date = pd.to_datetime(start_date)
    end_date = pd.to_datetime(end_date)

    result = attendance[
        (attendance["employee_id"] == employee_id) &
        (attendance["date"] >= start_date) &
        (attendance["date"] <= end_date)
    ].copy()

    return result
```

```python
result = get_attendance(
    employee_id="E0001",
    start_date="2026-08-01",
    end_date="2026-08-31"
)

display(result)
```

```python
def get_attendance_summary(
    employee_id,
    start_date,
    end_date
):
    """
    Calculate attendance statistics for an employee.
    """

    records = get_attendance(
        employee_id,
        start_date,
        end_date
    )

    summary = {
        "employee_id": employee_id,
        "start_date": start_date,
        "end_date": end_date,
        "office_days": int(
            (records["location"] == "office").sum()
        ),
        "home_days": int(
            (records["location"] == "home").sum()
        ),
        "business_trip_days": int(
            (records["location"] == "business_trip").sum()
        ),
        "sick_days": int(
            (records["status"] == "sick").sum()
        ),
        "leave_days": int(
            (records["status"] == "leave").sum()
        ),
        "missing_records": int(
            records["status"].isin(
                ["missing", "missing_checkout"]
            ).sum()
        )
    }

    return summary
```

```python
summary = get_attendance_summary(
    employee_id="E0001",
    start_date="2026-08-01",
    end_date="2026-08-31"
)

print(summary)
```

```python
def find_missing_attendance(
    employee_id,
    start_date,
    end_date
):
    """
    Find missing check-ins and check-outs
    for an employee.
    """

    records = get_attendance(
        employee_id,
        start_date,
        end_date
    )

    missing = records[
        records["status"].isin(
            ["missing", "missing_checkout"]
        )
    ].copy()

    return missing
```

```python
missing = find_missing_attendance(
    employee_id="E0001",
    start_date="2026-08-01",
    end_date="2026-08-31"
)

display(missing)
```

```python
def calculate_working_hours(
    employee_id,
    start_date,
    end_date
):
    """
    Calculate total recorded working hours
    for an employee within a date range.
    """

    records = get_attendance(
        employee_id,
        start_date,
        end_date
    )

    records = records[
        records["check_in"].notna() &
        records["check_out"].notna()
    ].copy()

    records["check_in_time"] = pd.to_datetime(
        records["check_in"],
        format="%H:%M"
    )

    records["check_out_time"] = pd.to_datetime(
        records["check_out"],
        format="%H:%M"
    )

    records["hours"] = (
        records["check_out_time"] -
        records["check_in_time"]
    ).dt.total_seconds() / 3600

    return round(records["hours"].sum(), 2)
```

```python
hours = calculate_working_hours(
    employee_id="E0001",
    start_date="2026-08-01",
    end_date="2026-08-31"
)

print(f"Total working hours: {hours}")
```

```python
def get_employee(employee_id):
    """
    Return employee information.
    """

    result = employees[
        employees["employee_id"] == employee_id
    ]

    if result.empty:
        return None

    return result.iloc[0].to_dict()
```

```python
employee = get_employee("E0001")

print(employee)
```

```python
def get_leave(
    employee_id,
    start_date=None,
    end_date=None
):
    """
    Return leave records for an employee.
    """

    result = leave[
        leave["employee_id"] == employee_id
    ].copy()

    if start_date is not None:
        start_date = pd.to_datetime(start_date)

        result = result[
            result["start_date"] >= start_date
        ]

    if end_date is not None:
        end_date = pd.to_datetime(end_date)

        result = result[
            result["end_date"] <= end_date
        ]

    return result
```

```python
employee_leave = get_leave(
    employee_id="E0001"
)

display(employee_leave)
```

```python
def get_leave_balance(employee_id):
    """
    Calculate remaining approved vacation days.
    """

    employee = get_employee(employee_id)

    if employee is None:
        return None

    vacation_entitlement = employee["vacation_days"]

    approved_leave = leave[
        (leave["employee_id"] == employee_id) &
        (leave["type"] == "vacation") &
        (leave["status"] == "approved")
    ]

    used_vacation = approved_leave["days"].sum()

    remaining = vacation_entitlement - used_vacation

    return {
        "employee_id": employee_id,
        "vacation_entitlement": int(vacation_entitlement),
        "vacation_used": int(used_vacation),
        "vacation_remaining": int(remaining)
    }
```

```python
balance = get_leave_balance("E0001")

print(balance)
```
